```python
# ============================================================
# 02_us_stock_analysis.ipynb
# 미국 주식 상승추세 스크리너
# ============================================================

# 처음 한 번만 실행
# %pip install -U pandas numpy matplotlib mplfinance


# --------------------------------------------------
# 1. 라이브러리
# --------------------------------------------------
from pathlib import Path
import platform

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import mplfinance as mpf

from IPython.display import display

pd.set_option("display.max_columns", 50)
pd.set_option("display.float_format", lambda x: f"{x:,.2f}")


# --------------------------------------------------
# 2. 한글 폰트
# --------------------------------------------------
if platform.system() == "Windows":
    KOREAN_FONT = "Malgun Gothic"
elif platform.system() == "Darwin":
    KOREAN_FONT = "AppleGothic"
else:
    KOREAN_FONT = "NanumGothic"

plt.rcParams["font.family"] = KOREAN_FONT
plt.rcParams["axes.unicode_minus"] = False
plt.rcParams["font.weight"] = "normal"
plt.rcParams["axes.titleweight"] = "normal"

print("운영체제 :", platform.system())
print("사용 폰트 :", KOREAN_FONT)


# --------------------------------------------------
# 3. 설정
# --------------------------------------------------
BASE_DIR = Path("us_stock_data")
STOCK_DIR = BASE_DIR / "stocks"
MARKET_DIR = BASE_DIR / "market"
META_DIR = BASE_DIR / "metadata"
SP500_FILE = META_DIR / "sp500_constituents.csv"

MIN_ROWS = 130
MIN_DOLLAR_VOLUME = 20_000_000

# 과열 기준
MAX_RET20 = 0.25
MAX_MA20_DIST = 0.12
MAX_DAY_GAIN5 = 0.15

# 균형형 기준
BAL_MOM_PCT = 0.70
BAL_RET20_MAX = 0.15
BAL_MA20_MIN = 0.02
BAL_MA20_MAX = 0.08
BAL_DAY_GAIN5 = 0.10
BAL_VOL_PCT = 0.70

MAX_STOP_LOSS = 0.08

TOP_N = 20

# 차트에 표시할 유형
# "균형형", "강한추세", "저과열", "일반", "급등주의", "과열주의"
CHART_TYPE = "균형형"
CHART_N = 10

print("데이터 폴더 :", BASE_DIR.resolve())
print("차트 유형   :", CHART_TYPE)


# --------------------------------------------------
# 4. 저장 파일 확인
# --------------------------------------------------
required_files = [
    SP500_FILE,
    MARKET_DIR / "SPY.csv",
    MARKET_DIR / "QQQ.csv"
]

missing = [str(file) for file in required_files if not file.exists()]

if missing:
    raise FileNotFoundError(
        "필수 파일이 없습니다.\n"
        + "\n".join(missing)
        + "\n먼저 01_us_stock_download.ipynb를 실행하세요."
    )

sp500 = pd.read_csv(SP500_FILE)

print("\nS&P 500 증권 수 :", len(sp500))
print("저장된 종목 CSV :", len(list(STOCK_DIR.glob("*.csv"))))

display(sp500)


# --------------------------------------------------
# 5. SPY / QQQ 시장 상태
# --------------------------------------------------
market_rows = []

for ticker in ["SPY", "QQQ"]:

    df = pd.read_csv(MARKET_DIR / f"{ticker}.csv", parse_dates=["Date"])
    df = df.sort_values("Date").drop_duplicates("Date").reset_index(drop=True)

    df["MA200"] = df["Close"].rolling(200).mean()

    # 6개월 전 ~ 1개월 전 수익률
    df["Momentum6_1M"] = df["Close"].shift(22) / df["Close"].shift(126) - 1

    last = df.iloc[-1]

    market_rows.append({
        "Ticker": ticker,
        "Date": last["Date"],
        "Close": last["Close"],
        "MA200": last["MA200"],
        "AboveMA200": last["Close"] > last["MA200"],
        "Momentum6_1M": last["Momentum6_1M"]
    })

market = pd.DataFrame(market_rows)

SPY_DATE = market.loc[market["Ticker"] == "SPY", "Date"].iloc[0]
SPY_MOM = market.loc[market["Ticker"] == "SPY", "Momentum6_1M"].iloc[0]
SPY_MARKET_OK = bool(market.loc[market["Ticker"] == "SPY", "AboveMA200"].iloc[0])

print("\n분석 기준일 :", SPY_DATE.date())
print("시장 상태   :", "상승장 - 신규 매수 허용" if SPY_MARKET_OK else "방어장 - 신규 매수 중단")

display(market)


# --------------------------------------------------
# 6. S&P 500 전체 종목 분석
# --------------------------------------------------
rows = []
errors = []

for _, info in sp500.iterrows():

    ticker = info["YahooTicker"]
    file = STOCK_DIR / f"{ticker}.csv"

    if not file.exists():
        errors.append([ticker, "파일없음"])
        continue

    try:
        df = pd.read_csv(file, parse_dates=["Date"])
        df = df.sort_values("Date").drop_duplicates("Date").reset_index(drop=True)
        df = df[df["Date"] <= SPY_DATE].copy()

        if len(df) < MIN_ROWS:
            errors.append([ticker, "데이터부족"])
            continue

        # 이동평균
        df["MA20"] = df["Close"].rolling(20).mean()
        df["MA60"] = df["Close"].rolling(60).mean()
        df["MA120"] = df["Close"].rolling(120).mean()

        # 수익률과 모멘텀
        df["Return1D"] = df["Close"].pct_change()
        df["Return20"] = df["Close"] / df["Close"].shift(20) - 1
        df["Momentum6_1M"] = df["Close"].shift(22) / df["Close"].shift(126) - 1

        # 이격, 거래대금, 변동성
        df["MA20Dist"] = df["Close"] / df["MA20"] - 1
        df["DollarVolume"] = df["Close"] * df["Volume"]
        df["AvgDollar20"] = df["DollarVolume"].rolling(20).mean()
        df["Volatility20"] = df["Return1D"].rolling(20).std()
        df["MaxDayGain5"] = df["Return1D"].rolling(5).max()

        # 이동평균 상승 여부
        df["MA20Up"] = df["MA20"] > df["MA20"].shift(5)
        df["MA60Up"] = df["MA60"] > df["MA60"].shift(5)
        df["MA120Up"] = df["MA120"] > df["MA120"].shift(5)

        last = df.iloc[-1]

        # 시장 기준일과 날짜가 다른 종목 제외
        if last["Date"].date() != SPY_DATE.date():
            errors.append([ticker, "최신일불일치"])
            continue

        # 상승추세
        trend = (
            last["Close"] > last["MA20"] > last["MA60"] > last["MA120"]
            and last["MA20Up"]
            and last["MA60Up"]
            and last["MA120Up"]
        )

        relative_momentum = last["Momentum6_1M"] - SPY_MOM

        rows.append({
            "Ticker": ticker,
            "Name": info["Name"],
            "Sector": info["Sector"],
            "Industry": info["Industry"],
            "Date": last["Date"],
            "Close": last["Close"],
            "MA20": last["MA20"],
            "MA60": last["MA60"],
            "MA120": last["MA120"],
            "Momentum6_1M": last["Momentum6_1M"],
            "RelativeMomentum": relative_momentum,
            "Return20": last["Return20"],
            "MA20Dist": last["MA20Dist"],
            "MaxDayGain5": last["MaxDayGain5"],
            "AvgDollar20": last["AvgDollar20"],
            "Volatility20": last["Volatility20"],
            "Trend": trend
        })

    except Exception as e:
        errors.append([ticker, str(e)[:80]])

result = pd.DataFrame(rows)

print("\n분석 성공 :", len(result))
print("제외/오류 :", len(errors))

if errors:
    error_df = pd.DataFrame(errors, columns=["Ticker", "Reason"])
    display(error_df.head(20))


# --------------------------------------------------
# 7. 기본 필터
# --------------------------------------------------
result["AbsMomentumOK"] = result["Momentum6_1M"] > 0
result["RelativeMomentumOK"] = result["RelativeMomentum"] > 0
result["LiquidityOK"] = result["AvgDollar20"] >= MIN_DOLLAR_VOLUME

result["OverheatOK"] = (
    (result["Return20"] <= MAX_RET20)
    & (result["MA20Dist"] <= MAX_MA20_DIST)
    & (result["MaxDayGain5"] <= MAX_DAY_GAIN5)
)

result["ScreenOK"] = (
    result["Trend"]
    & result["AbsMomentumOK"]
    & result["RelativeMomentumOK"]
    & result["LiquidityOK"]
    & result["OverheatOK"]
)

print("\n전체 분석 종목 :", len(result))
print("추세 통과      :", result["Trend"].sum())
print("최종 필터 통과 :", result["ScreenOK"].sum())

filter_summary = pd.DataFrame({
    "조건": [
        "MA 상승추세",
        "6-1M > 0",
        "SPY 상대강도 > 0",
        "20일 평균 거래대금 >= $20M",
        "과열 필터 통과",
        "전체 조건 통과"
    ],
    "종목수": [
        result["Trend"].sum(),
        result["AbsMomentumOK"].sum(),
        result["RelativeMomentumOK"].sum(),
        result["LiquidityOK"].sum(),
        result["OverheatOK"].sum(),
        result["ScreenOK"].sum()
    ]
})

display(filter_summary)


# --------------------------------------------------
# 8. BuyScore
# --------------------------------------------------
candidates = result[result["ScreenOK"]].copy()

if candidates.empty:
    raise RuntimeError("현재 조건을 모두 통과한 종목이 없습니다.")

# 후보 종목 안에서 percentile 계산
candidates["MomentumPct"] = candidates["Momentum6_1M"].rank(pct=True)
candidates["LiquidityPct"] = candidates["AvgDollar20"].rank(pct=True)
candidates["VolatilityPct"] = candidates["Volatility20"].rank(pct=True)

# 100점 만점
candidates["MomentumScore"] = candidates["MomentumPct"] * 45

# MA20보다 약 +5% 위를 가장 선호
candidates["DistanceScore"] = (
    1 - abs(candidates["MA20Dist"] - 0.05) / 0.07
).clip(0, 1) * 30

# 변동성이 낮을수록 높은 점수
candidates["StabilityScore"] = (1 - candidates["VolatilityPct"]) * 20

# S&P 500은 대부분 유동성이 높으므로 5점만 부여
candidates["LiquidityScore"] = candidates["LiquidityPct"] * 5

candidates["BuyScore"] = (
    candidates["MomentumScore"]
    + candidates["DistanceScore"]
    + candidates["StabilityScore"]
    + candidates["LiquidityScore"]
)


# --------------------------------------------------
# 9. 유형 분류
# --------------------------------------------------
balanced = (
    (candidates["MomentumPct"] >= BAL_MOM_PCT)
    & candidates["Return20"].between(0, BAL_RET20_MAX)
    & candidates["MA20Dist"].between(BAL_MA20_MIN, BAL_MA20_MAX)
    & (candidates["MaxDayGain5"] <= BAL_DAY_GAIN5)
    & (candidates["VolatilityPct"] <= BAL_VOL_PCT)
)

surge = (
    (candidates["MaxDayGain5"] > 0.10)
    | (candidates["Return20"] > 0.20)
)

overheated = candidates["MA20Dist"] > 0.09

strong = (
    (candidates["MomentumPct"] >= 0.85)
    & (candidates["MA20Dist"] <= 0.10)
)

low_heat = (
    (candidates["MA20Dist"] < 0.03)
    & (candidates["Return20"] < 0.08)
)

# 앞에 있는 조건이 우선 적용됨
candidates["Type"] = np.select(
    [surge, overheated, balanced, strong, low_heat],
    ["급등주의", "과열주의", "균형형", "강한추세", "저과열"],
    default="일반"
)

print("\n유형별 종목 수")
display(candidates["Type"].value_counts().to_frame("종목수"))


# --------------------------------------------------
# 10. 손절가 / 1R / 2R
# --------------------------------------------------
candidates["EntryPrice"] = candidates["Close"]

# MA60 또는 -8% 중 높은 가격을 손절가로 사용
candidates["StopPrice"] = np.maximum(
    candidates["MA60"],
    candidates["EntryPrice"] * (1 - MAX_STOP_LOSS)
)

candidates["R"] = candidates["EntryPrice"] - candidates["StopPrice"]
candidates["Target1R"] = candidates["EntryPrice"] + candidates["R"]
candidates["Target2R"] = candidates["EntryPrice"] + 2 * candidates["R"]

candidates["CurrentStage"] = "신규후보" if SPY_MARKET_OK else "시장필터OFF"
candidates["SellSignal"] = np.where(candidates["Close"] < candidates["MA20"], "MA20 이탈", "없음")
candidates["BuyAllowed"] = "허용" if SPY_MARKET_OK else "중단"


# --------------------------------------------------
# 11. BuyScore 순위
# --------------------------------------------------
candidates = candidates.sort_values("BuyScore", ascending=False).reset_index(drop=True)
candidates["Rank"] = np.arange(1, len(candidates) + 1)

# 보기 편한 단위
candidates["Momentum6_1M_%"] = candidates["Momentum6_1M"] * 100
candidates["RelativeMomentum_%"] = candidates["RelativeMomentum"] * 100
candidates["Return20_%"] = candidates["Return20"] * 100
candidates["MA20Dist_%"] = candidates["MA20Dist"] * 100
candidates["Volatility20_%"] = candidates["Volatility20"] * 100
candidates["AvgDollar20_M"] = candidates["AvgDollar20"] / 1_000_000

show_cols = [
    "Rank", "Ticker", "Name", "Sector", "Type", "BuyScore",
    "Momentum6_1M_%", "RelativeMomentum_%",
    "Return20_%", "MA20Dist_%", "Volatility20_%",
    "AvgDollar20_M", "BuyAllowed"
]

print(f"\n전체 TOP {min(TOP_N, len(candidates))}")
display(candidates[show_cols].head(TOP_N).round(2))


# --------------------------------------------------
# 12. 균형형 종목
# --------------------------------------------------
balanced_result = candidates[candidates["Type"] == "균형형"].copy()

balanced_cols = [
    "Rank", "Ticker", "Name", "Sector", "BuyScore",
    "Momentum6_1M_%", "RelativeMomentum_%",
    "Return20_%", "MA20Dist_%", "Volatility20_%"
]

print("\n균형형 종목 :", len(balanced_result))
display(balanced_result[balanced_cols].head(TOP_N).round(2))


# --------------------------------------------------
# 13. 매매 계획
# --------------------------------------------------
trade_plan = candidates[
    [
        "Rank", "Ticker", "Name", "Type", "BuyScore",
        "EntryPrice", "StopPrice", "Target1R", "Target2R",
        "CurrentStage", "SellSignal"
    ]
].head(TOP_N).copy()

trade_plan = trade_plan.rename(columns={
    "EntryPrice": "매수가",
    "StopPrice": "손절가",
    "Target1R": "1R(30%매도)",
    "Target2R": "2R(30%매도)",
    "CurrentStage": "현재단계",
    "SellSignal": "매도신호"
})

price_cols = ["매수가", "손절가", "1R(30%매도)", "2R(30%매도)"]
trade_plan[price_cols] = trade_plan[price_cols].round(2)

print("\n매매 계획")
display(trade_plan)


# --------------------------------------------------
# 14. 결과 저장
# --------------------------------------------------
OUTPUT_FILE = BASE_DIR / "us_stock_analysis_result.csv"
candidates.to_csv(OUTPUT_FILE, index=False)

print("\n분석 결과 저장 :", OUTPUT_FILE)


# --------------------------------------------------
# 15. mplfinance 한글 스타일
# --------------------------------------------------
mpf_style = mpf.make_mpf_style(
    base_mpf_style="yahoo",
    rc={
        "font.family": KOREAN_FONT,
        "font.weight": "normal",
        "axes.titleweight": "normal",
        "axes.unicode_minus": False
    }
)


# --------------------------------------------------
# 16. 원하는 유형의 종목만 캔들 차트
# --------------------------------------------------
valid_types = ["균형형", "강한추세", "저과열", "일반", "급등주의", "과열주의"]

if CHART_TYPE not in valid_types:
    raise ValueError(f"CHART_TYPE은 다음 중 하나여야 합니다: {valid_types}")

chart_result = (
    candidates[candidates["Type"] == CHART_TYPE]
    .sort_values("BuyScore", ascending=False)
    .head(CHART_N)
)

print(f"\n[{CHART_TYPE}] 차트 대상 :", len(chart_result))

if chart_result.empty:
    print(f"{CHART_TYPE} 유형의 종목이 없습니다.")

else:
    display(
        chart_result[
            ["Rank", "Ticker", "Name", "Sector", "BuyScore",
             "Momentum6_1M_%", "Return20_%", "MA20Dist_%"]
        ].round(2)
    )

    chart_tickers = chart_result["Ticker"].tolist()
    print("차트 종목 :", chart_tickers)

    for ticker in chart_tickers:

        df = pd.read_csv(STOCK_DIR / f"{ticker}.csv", parse_dates=["Date"])
        df = df[df["Date"] <= SPY_DATE].sort_values("Date").tail(120).copy()
        df = df.set_index("Date")

        info = candidates[candidates["Ticker"] == ticker].iloc[0]
        title = f"{ticker} | {info['Type']} | BuyScore {info['BuyScore']:.1f}"

        fig, axes = mpf.plot(
            df[["Open", "High", "Low", "Close", "Volume"]],
            type="candle",
            mav=(20, 60, 120),
            volume=True,
            style=mpf_style,
            figsize=(12, 7),
            tight_layout=True,
            returnfig=True
        )

        fig.suptitle(
            title,
            fontsize=14,
            fontfamily=KOREAN_FONT,
            fontweight="normal",
            y=0.99
        )

        plt.show()
```
