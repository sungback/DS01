```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic.csv')
df
```

```python
df.info()
# 1. sex(글자) ==> 숫자로 변경, 예: female(여성)은 1로, male(남성)은 0으로 변환
# 2. age의 결측치 263개를 채우자.
# 3. fare의 결측치 1개는 버리자.
```

```python
# sex(성별) --> 숫자
# df['추가할_컬럼명'] = 계산식
df['sex2'] = df['sex'].map( {'female':1, 'male':0} ) # 1줄
df
```

```python
# 운임의 결측치를 삭제
df = df.dropna(subset='fare') # 운임 컬럼의 결측치를 삭제한다.
df # 1309 - 1 ==> 1308
```

```python
# age(나이)의 결측치를 채우기
```

```python
df.isnull().sum()
```

```python
# 호칭을 뽑는다. 예) Miss. --> Miss, 영어 대소문자가 1번이상 나오고 점(.)이 있으면
df['title'] = df['name'].str.extract("([A-Za-z]+)\.") # 정규표현식 활용
df
```

```python
df['title'].value_counts()
```

```python
# 호칭별 나이의 중앙값 얻기
val = df.groupby('title')['age'].transform('median')
val
```

```python
# 호칭별 나이의 결측치를 채우기
df['age'] = df['age'].fillna(val)
```

```python
# 1. 호칭 뽑기 : df['title'] = df['name'].str.extract("([A-Za-z]+)\.")
# 2. 호칭별 중앙값 얻기 : val = df.groupby('title')['age'].transform('median')
# 3. 나이의 결측치를 채우기 : df['age'] = df['age'].fillna(val)
```

```python
df.isnull().sum() # age    0 <== 나이의 결측치 채우기 성공!
```

```python
# 저장
df.to_csv('titanic_ok.csv', index=False) # 인덱스는 빼고 저장
```
